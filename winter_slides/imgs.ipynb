```python
import matplotlib.pyplot as plt

# Данные
models = [
    {"name": "LLaMA 1", "year": 2023, "size": 13},
    {"name": "LLaMA 2", "year": 2023, "size": 70},
    {"name": "LLaMA 3.1", "year": 2024, "size": 405},
    {"name": "Qwen 7B", "year": 2023, "size": 7},
    {"name": "Qwen 2.5-Max", "year": 2025, "size": 72},
    {"name": "DeepSeek-LLM 7B", "year": 2023, "size": 7},
    {"name": "DeepSeek-LLM 67B", "year": 2023, "size": 67},
    {"name": "Mistral 7B", "year": 2023, "size": 7},
    {"name": "Mixtral 8x7B", "year": 2023, "size": 56},
    {"name": "Grok-1", "year": 2023, "size": 314},
]

# Извлечение данных для построения графика
years = [model["year"] for model in models]
sizes = [model["size"] for model in models]
names = [model["name"] for model in models]

# Построение графика
plt.figure(figsize=(10, 6))
plt.scatter(years, sizes, color='blue')

# Добавление подписей к точкам
for i, name in enumerate(names):
    plt.text(years[i], sizes[i], name, fontsize=9, ha='right')

plt.title('Год выпуска vs. Размер модели (млрд параметров)')
plt.xlabel('Год выпуска')
plt.ylabel('Размер модели (млрд параметров)')
plt.grid(True)
plt.show()
```
